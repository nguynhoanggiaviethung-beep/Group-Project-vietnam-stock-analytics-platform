```python
def load_one_file(file_path, exchange):
   
    # ---------------------------------------------------------
    # 1. Kiểm tra file có tồn tại hay không
    # ---------------------------------------------------------
    if not file_path.exists():
        raise FileNotFoundError(
            f"Không tìm thấy file: {file_path}"
        )

    # ---------------------------------------------------------
    # 2. Một số file CSV có thể sử dụng encoding khác nhau.
    #    Vì vậy, ta thử lần lượt một số encoding phổ biến.
    # ---------------------------------------------------------
    encodings = [
        "utf-8-sig",
        "utf-8",
        "cp1252"
    ]

    df = None
    last_error = None

    for encoding in encodings:
        try:
            df = pd.read_csv(
                file_path,
                encoding=encoding
            )

            print(
                f"Đọc thành công: {file_path.name} "
                f"| encoding = {encoding}"
            )

            break

        except UnicodeDecodeError as e:
            last_error = e

    # ---------------------------------------------------------
    # 3. Nếu tất cả encoding đều thất bại
    # ---------------------------------------------------------
    if df is None:
        raise UnicodeDecodeError(
            "Không thể đọc file bằng các encoding đã thử."
        ) from last_error

    # ---------------------------------------------------------
    # 4. Loại bỏ khoảng trắng thừa ở tên cột.
    #
    #    Ví dụ:
    #    " <Ticker> " → "<Ticker>"
    # ---------------------------------------------------------
    df.columns = (
        df.columns
        .astype(str)
        .str.strip()
    )

    # ---------------------------------------------------------
    # 5. Thêm cột exchange dựa trên nguồn file.
    #
    #    HNX file   → HNX
    #    UPCOM file → UPCOM
    #    HSX file   → HOSE
    # ---------------------------------------------------------
    df["exchange"] = exchange

    # ---------------------------------------------------------
    # 6. In thông tin cơ bản để kiểm tra dữ liệu.
    # ---------------------------------------------------------
    print(f"Số dòng: {len(df):,}")
    print(f"Số cột: {len(df.columns)}")

    return df
```

```python
def apply_schema(df):
    """
    Chuẩn hóa Raw DataFrame từ cấu trúc CafeF
    về schema thống nhất của project.

    Schema đầu ra:
    ticker
    exchange
    date
    open
    high
    low
    close
    volume
    """

    # ---------------------------------------------------------
    # 1. Khai báo mapping giữa tên cột gốc của CafeF
    #    và tên cột chuẩn của project
    # ---------------------------------------------------------
    column_mapping = {
        "<Ticker>": "ticker",
        "<DTYYYYMMDD>": "date",
        "<Open>": "open",
        "<High>": "high",
        "<Low>": "low",
        "<Close>": "close",
        "<Volume>": "volume",
        "exchange": "exchange"
    }

    # ---------------------------------------------------------
    # 2. Kiểm tra xem Raw DataFrame có đủ các cột cần thiết
    #    hay không
    # ---------------------------------------------------------
    required_columns = list(column_mapping.keys())

    missing_columns = [
        col for col in required_columns
        if col not in df.columns
    ]

    if missing_columns:
        raise ValueError(
            f"Thiếu các cột cần thiết: {missing_columns}"
        )

    # ---------------------------------------------------------
    # 3. Chỉ lấy những cột cần thiết cho project
    # ---------------------------------------------------------
    df = df[required_columns].copy()

    # ---------------------------------------------------------
    # 4. Đổi tên cột theo schema chuẩn
    # ---------------------------------------------------------
    df = df.rename(columns=column_mapping)

    # ---------------------------------------------------------
    # 5. Chuẩn hóa ticker và exchange thành kiểu string
    # ---------------------------------------------------------
    df["ticker"] = df["ticker"].astype("string")
    df["exchange"] = df["exchange"].astype("string")

    # ---------------------------------------------------------
    # 6. Chuyển date từ dạng YYYYMMDD
    #    Ví dụ:
    #    20260903 → 2026-09-03
    #
    #    errors="coerce" giúp giá trị ngày không hợp lệ
    #    chuyển thành NaT thay vì làm chương trình bị dừng.
    # ---------------------------------------------------------
    df["date"] = pd.to_datetime(
        df["date"],
        format="%Y%m%d",
        errors="coerce"
    )

    # ---------------------------------------------------------
    # 7. Chuyển các biến giá sang kiểu số
    # ---------------------------------------------------------
    price_columns = [
        "open",
        "high",
        "low",
        "close"
    ]

    for col in price_columns:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        ).astype("float64")

    # ---------------------------------------------------------
    # 8. Chuyển volume sang kiểu số
    #
    #    Dùng Int64 dạng nullable của Pandas để vẫn có thể
    #    chứa giá trị thiếu (NA) nếu xuất hiện.
    # ---------------------------------------------------------
    df["volume"] = pd.to_numeric(
        df["volume"],
        errors="coerce"
    ).astype("Int64")

    # ---------------------------------------------------------
    # 9. Sắp xếp đúng thứ tự schema của project
    # ---------------------------------------------------------
    schema = [
        "ticker",
        "exchange",
        "date",
        "open",
        "high",
        "low",
        "close",
        "volume"
    ]

    df = df[schema]

    return df
```

```python
def get_raw_data(base_dir=BASE_DIR):
    """
    Thực hiện toàn bộ quy trình Data Acquisition:

    1. Đọc 3 file CSV local của CafeF
    2. Gán exchange cho từng file
    3. Gộp dữ liệu thành một DataFrame
    4. Chuẩn hóa schema
    5. Trả về DataFrame hoàn chỉnh
    """

    # ---------------------------------------------------------
    # Bước 1: Đọc dữ liệu từ 3 file CSV
    # ---------------------------------------------------------
    raw_df = load_local_data(base_dir)

    # ---------------------------------------------------------
    # Bước 2: Chuẩn hóa schema
    # ---------------------------------------------------------
    standardized_df = apply_schema(raw_df)

    # ---------------------------------------------------------
    # Bước 3: Trả về DataFrame
    # ---------------------------------------------------------
    return standardized_df
```

```python
def load_local_data(base_dir=BASE_DIR):
    """
    Đọc đúng 3 file CSV local của CafeF,
    gán exchange và gộp thành một Raw DataFrame.
    """

    dataframes = []

    # ---------------------------------------------------------
    # Duyệt qua 3 file HNX, UPCOM và HOSE
    # ---------------------------------------------------------
    for exchange, filename in [
        ("HNX", "CafeF.HNX.Upto03.09.2026.csv"),
        ("UPCOM", "CafeF.UPCOM.Upto03.09.2026.csv"),
        ("HOSE", "CafeF.HSX.Upto03.09.2026.csv")
    ]:

        # Tạo đường dẫn đầy đủ tới file
        file_path = base_dir / filename

        # Đọc file và thêm exchange
        df = load_one_file(
            file_path,
            exchange
        )

        # Lưu DataFrame vào danh sách
        dataframes.append(df)

    # ---------------------------------------------------------
    # Gộp 3 DataFrame thành một DataFrame duy nhất
    # ---------------------------------------------------------
    raw_df = pd.concat(
        dataframes,
        ignore_index=True,
        sort=False
    )

    return raw_df
```

```python
standardized_df = get_raw_data()

print("Data Acquisition hoàn tất!")
print("Kích thước dữ liệu:", standardized_df.shape)
```
